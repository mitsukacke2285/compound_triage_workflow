```python
from rdkit import Chem

# Output file
output_smi = "top_5_AS.smi"

# Conversion of SDF into SMI
def sdf_into_smiles(comp):
    supplier = Chem.SDMolSupplier(str(comp))
    compounds = []

    # Save as a list of SMILES
    for i, mol in enumerate(supplier):
        if mol is None:
            continue
        smiles = Chem.MolToSmiles(mol, canonical=True)
        compounds.append(smiles)

    # Save as SMI file
    with open(output_smi, 'w') as f:
        for smi in compounds:
            f.write(smi + "\n")
            
    return compounds

# Input file
input_sdf = 'top_5_AS.sdf'
sdf_into_smiles(input_sdf)
```
