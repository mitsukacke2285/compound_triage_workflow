```python
import config
from rdkit import Chem

# Output file
output_smiles = config.OUTPUT_SMILES

# Conversion of SDF into SMI
def sdf_into_smiles(comp):
    supplier = Chem.SDMolSupplier(str(comp))
    compounds = []

    # Save as a list of SMILES
    for i, mol in enumerate(supplier):
        if mol is None:
            continue
        smiles = Chem.MolToSmiles(mol, canonical=True)
        compounds.append(smiles)

    # Save as SMI file
    with open(output_smiles, 'w') as f:
        for smi in compounds:
            f.write(smi + "\n")
            
    return compounds

# Input file
#input_sdf = '3000er_no_PAINS.sdf'
input_sdf = config.INPUT_SDF
sdf_into_smiles(input_sdf)

print(f'{input_sdf} has been successfully loaded and saved as {output_smiles}!')
```
