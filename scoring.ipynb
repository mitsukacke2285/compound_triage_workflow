```python
import pandas as pd
import numpy as np

### Load csv file ###
profiled_compounds = 'standardized_smiles_admet.csv'
df = pd.read_csv(profiled_compounds)

#######################################
### Filter out PAINS, Brenk and NIH ###
#######################################
def structural_filter():
    conditions = (
        (df['PAINS'] == True) |
        (df['BRENK'] == True) |
        (df['NIH'] == True)
    )

    indices_drop = df[conditions].index
    df_filtered = df.drop(indices_drop, inplace=False)
    
    return df_filtered

structural_filter()
```
